```python
# save as pinn_vae_compare.py
import os
import math
import random
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.optim as optim

# ---------------------
# Basic settings
# ---------------------
seed = 42
random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
results_dir = "results"
os.makedirs(results_dir, exist_ok=True)

# ---------------------
# Data: generate many SHO trajectories
# x(t) = A * cos(w t + phi)
# We produce N trajectories, each sampled at T equally spaced timesteps.
# ---------------------
def generate_sho_dataset(N=2000, T=128, t_max=2*math.pi, noise_std=0.0):
    ts = np.linspace(0.0, t_max, T)
    data = np.zeros((N, T), dtype=np.float32)
    params = np.zeros((N, 3), dtype=np.float32)  # A, w, phi
    for i in range(N):
        A = np.random.uniform(0.5, 2.0)            # amplitude
        w = np.random.uniform(0.8, 2.5)            # frequency
        phi = np.random.uniform(-math.pi, math.pi) # phase
        traj = A * np.cos(w * ts + phi)
        traj = traj + np.random.normal(scale=noise_std, size=traj.shape)
        data[i] = traj.astype(np.float32)
        params[i] = np.array([A, w, phi], dtype=np.float32)
    return data, params, ts

# ---------------------
# Small MLP VAE building blocks
# ---------------------
class Encoder(nn.Module):
    def __init__(self, input_dim, hidden_dim, latent_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
        )
        self.mu = nn.Linear(hidden_dim, latent_dim)
        self.logvar = nn.Linear(hidden_dim, latent_dim)
    def forward(self, x):
        h = self.net(x)
        return self.mu(h), self.logvar(h)

class ContextEncoder(nn.Module):
    def __init__(self, context_dim, latent_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(context_dim, 64),
            nn.ReLU(),
            nn.Linear(64, latent_dim)
        )
    def forward(self, c):
        return self.net(c)

class Decoder(nn.Module):
    def __init__(self, latent_dim, hidden_dim, output_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(latent_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, output_dim),
        )
    def forward(self, z):
        return self.net(z)

# ---------------------
# Reparameterization utility
# ---------------------
def reparameterize(mu, logvar):
    std = torch.exp(0.5 * logvar)
    eps = torch.randn_like(std)
    return mu + eps * std

# ---------------------
# Physics loss for SHO focused on known params
# uses x'' + w^2 x = 0 for pure SHO, implemented softly by
# comparing decoded signal to analytic SHO using parameters (A,w,phi).
# For CAR we use context params. For PII we optionally use predicted params.
# ---------------------
def sho_reconstruction_from_params(params, ts):
    # params: [batch, 3] -> A,w,phi
    A = params[:, 0:1]  # [B,1]
    w = params[:, 1:2]
    phi = params[:, 2:3]
    t = torch.tensor(ts, dtype=torch.float32, device=params.device).unsqueeze(0)  # [1, T]
    return A * torch.cos(w * t + phi)

# ---------------------
# Training utilities: single epoch loops for both methods
# ---------------------
def train_car_pinn_vae(data, params, ts, epochs=200, batch_size=128, latent_dim=4, lr=1e-3):
    """
    CAR + PINN-VAE.
    Context: true params A,w,phi are fed into context encoder.
    Decoder receives z + context_emb.
    Physics loss encourages recon to match analytic SHO from context params.
    """
    N, T = data.shape
    input_dim = T
    context_dim = params.shape[1]  # 3: A,w,phi
    hidden_dim = 256

    enc = Encoder(input_dim, hidden_dim, latent_dim).to(device)
    ctx_enc = ContextEncoder(context_dim, latent_dim).to(device)
    dec = Decoder(latent_dim, hidden_dim, input_dim).to(device)

    optim_all = optim.Adam(list(enc.parameters()) + list(ctx_enc.parameters()) + list(dec.parameters()), lr=lr)
    mse = nn.MSELoss()

    dataset = torch.utils.data.TensorDataset(torch.from_numpy(data), torch.from_numpy(params))
    loader = torch.utils.data.DataLoader(dataset, batch_size=batch_size, shuffle=True)

    for ep in range(epochs):
        total_loss = 0.0
        for xb, pb in loader:
            xb = xb.to(device)
            pb = pb.to(device)
            mu, logvar = enc(xb)
            z = reparameterize(mu, logvar)
            ctx_emb = ctx_enc(pb)
            recon = dec(z + ctx_emb)  # additive CAR

            # losses
            recon_loss = mse(recon, xb)
            kld = -0.5 * torch.mean(torch.sum(1 + logvar - mu.pow(2) - logvar.exp(), dim=1))
            # physics loss matches recon to known SHO using context params
            sho_target = sho_reconstruction_from_params(pb, ts)
            phy_loss = mse(recon, sho_target)

            loss = recon_loss + 1e-3 * kld + 1.0 * phy_loss
            optim_all.zero_grad()
            loss.backward()
            optim_all.step()
            total_loss += loss.item() * xb.size(0)
        total_loss /= N
        if (ep % 50) == 0 or ep == epochs - 1:
            print(f"[CAR] Epoch {ep:03d} Loss {total_loss:.6f}")

    # After training, return models and some diagnostics
    return enc, ctx_enc, dec

def train_pii_pinn_vae(data, params, ts, epochs=200, batch_size=128, latent_dim=4, lr=1e-3):
    """
    PII + PINN-VAE.
    No context fed to decoder. We force identifiability by adding an
    intervention-consistency / identification loss that trains a small linear probe
    from z to true parameters. That probe is trained jointly to encourage latents
    to align with parameters that were varied across the dataset.
    Physics loss uses analytic signal built from the probe predictions optionally.
    """
    N, T = data.shape
    input_dim = T
    hidden_dim = 256

    enc = Encoder(input_dim, hidden_dim, latent_dim).to(device)
    dec = Decoder(latent_dim, hidden_dim, input_dim).to(device)
    # linear probe from latent to physical params
    probe = nn.Linear(latent_dim, params.shape[1]).to(device)

    optim_all = optim.Adam(list(enc.parameters()) + list(dec.parameters()) + list(probe.parameters()), lr=lr)
    mse = nn.MSELoss()

    dataset = torch.utils.data.TensorDataset(torch.from_numpy(data), torch.from_numpy(params))
    loader = torch.utils.data.DataLoader(dataset, batch_size=batch_size, shuffle=True)

    for ep in range(epochs):
        total_loss = 0.0
        for xb, pb in loader:
            xb = xb.to(device)
            pb = pb.to(device)
            mu, logvar = enc(xb)
            z = reparameterize(mu, logvar)
            recon = dec(z)  # no context

            # losses
            recon_loss = mse(recon, xb)
            kld = -0.5 * torch.mean(torch.sum(1 + logvar - mu.pow(2) - logvar.exp(), dim=1))

            # identification / PII-style loss: linear probe should predict true params
            pred_params = probe(mu)  # note using mu (deterministic encodings) for stability
            id_loss = mse(pred_params, pb)

            # physics loss can be enforced using predicted params from probe
            sho_target_from_probe = sho_reconstruction_from_params(pred_params, ts)
            phy_loss = mse(recon, sho_target_from_probe)

            # combine. weighting of id_loss is crucial. you can sweep it.
            loss = recon_loss + 1e-3 * kld + 1.0 * id_loss + 1.0 * phy_loss
            optim_all.zero_grad()
            loss.backward()
            optim_all.step()
            total_loss += loss.item() * xb.size(0)
        total_loss /= N
        if (ep % 50) == 0 or ep == epochs - 1:
            print(f"[PII] Epoch {ep:03d} Loss {total_loss:.6f}")

    return enc, dec, probe

# ---------------------
# Small eval / visualization utilities
# ---------------------
def plot_reconstruction_sample(model_components, data, params, ts, which="CAR", n=6):
    """
    model_components: tuples depending on method
    """
    idx = np.random.choice(len(data), size=n, replace=False)
    fig, axes = plt.subplots(n, 1, figsize=(6, 1.2 * n), sharex=True)
    for i, ax in enumerate(axes):
        x = torch.from_numpy(data[idx[i:i+1]]).to(device)
        p = torch.from_numpy(params[idx[i:i+1]]).to(device)
        if which == "CAR":
            enc, ctx_enc, dec = model_components
            mu, logvar = enc(x)
            z = reparameterize(mu, logvar)
            ctx_emb = ctx_enc(p)
            recon = dec(z + ctx_emb).detach().cpu().numpy()[0]
        else:
            enc, dec, probe = model_components
            mu, logvar = enc(x)
            z = reparameterize(mu, logvar)
            recon = dec(z).detach().cpu().numpy()[0]

        ax.plot(ts, data[idx[i]], label="true", linewidth=1)
        ax.plot(ts, recon, label="recon", linewidth=1)
        ax.set_ylabel(f"traj {idx[i]}")
        ax.legend(loc="upper right", fontsize="small")
    plt.tight_layout()
    fname = os.path.join(results_dir, f"recon_{which}.png")
    plt.savefig(fname)
    plt.close()
    print(f"Saved {fname}")

def plot_latent_vs_params(enc, ctx_enc, dec, data, params, method="CAR"):
    # encode full dataset and scatter latent dims vs params
    enc.eval()
    with torch.no_grad():
        X = torch.from_numpy(data).to(device)
        mu, _ = enc(X)
        Z = mu.cpu().numpy()  # [N, latent_dim]
    P = params  # numpy [N,3]
    # make scatter plots for first 3 latent dims vs three params
    fig, axes = plt.subplots(3, 3, figsize=(9, 9))
    for i in range(3):
        for j in range(3):
            axes[i, j].scatter(P[:, j], Z[:, i], s=6, alpha=0.6)
            axes[i, j].set_xlabel(["A", "w", "phi"][j])
            axes[i, j].set_ylabel(f"z{i}")
    plt.tight_layout()
    fname = os.path.join(results_dir, f"latent_vs_params_{method}.png")
    plt.savefig(fname)
    plt.close()
    print(f"Saved {fname}")

# ---------------------
# Main run: make dataset, train both methods, plot
# ---------------------
if __name__ == "__main__":
    # dataset with variation over A,w,phi
    data, params, ts = generate_sho_dataset(N=2000, T=128, t_max=2*math.pi, noise_std=0.02)
    print("Dataset shapes", data.shape, params.shape)

    # CAR experiment
    enc_car, ctx_enc, dec_car = train_car_pinn_vae(data, params, ts, epochs=300, batch_size=256, latent_dim=4, lr=1e-3)
    plot_reconstruction_sample((enc_car, ctx_enc, dec_car), data, params, ts, which="CAR", n=6)
    plot_latent_vs_params(enc_car, ctx_enc, dec_car, data, params, method="CAR")

    # PII experiment
    enc_pii, dec_pii, probe = train_pii_pinn_vae(data, params, ts, epochs=300, batch_size=256, latent_dim=4, lr=1e-3)
    plot_reconstruction_sample((enc_pii, dec_pii, probe), data, params, ts, which="PII", n=6)
    # scatter latents vs params for PII: use mu from enc_pii
    enc_pii.eval()
    with torch.no_grad():
        X = torch.from_numpy(data).to(device)
        mu_pii, _ = enc_pii(X)
        Zp = mu_pii.cpu().numpy()
    fig, axes = plt.subplots(3, 3, figsize=(9, 9))
    for i in range(3):
        for j in range(3):
            axes[i, j].scatter(params[:, j], Zp[:, i], s=6, alpha=0.6)
            axes[i, j].set_xlabel(["A", "w", "phi"][j])
            axes[i, j].set_ylabel(f"z{i}")
    plt.tight_layout()
    fname = os.path.join(results_dir, "latent_vs_params_PII.png")
    plt.savefig(fname)
    plt.close()
    print(f"Saved {fname}")

    # also inspect linear probe performance on held-out data for PII
    with torch.no_grad():
        X = torch.from_numpy(data).to(device)
        mu_pii, _ = enc_pii(X)
        pred_params = probe(mu_pii).cpu().numpy()
    mse_probe = np.mean((pred_params - params)**2)
    print(f"PII probe MSE on train set {mse_probe:.6f}")
```

# Claude Codes

## Context-Aware PINN-VAE

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
from torch.utils.data import Dataset, DataLoader
import matplotlib.pyplot as plt
from pathlib import Path

Path("results").mkdir(exist_ok=True)

class SHODataset(Dataset):
    def __init__(self, n_samples=5000, t_points=100, t_max=10.0):
        self.n_samples = n_samples
        self.t = np.linspace(0, t_max, t_points)
        
        self.amplitudes = np.random.uniform(0.5, 3.0, n_samples)
        self.frequencies = np.random.uniform(0.5, 3.0, n_samples)
        self.phases = np.random.uniform(0, 2*np.pi, n_samples)
        
        self.trajectories = []
        for i in range(n_samples):
            A, w, phi = self.amplitudes[i], self.frequencies[i], self.phases[i]
            x = A * np.sin(w * self.t + phi)
            self.trajectories.append(x)
        
        self.trajectories = np.array(self.trajectories, dtype=np.float32)
        
    def __len__(self):
        return self.n_samples
    
    def __getitem__(self, idx):
        return {
            'trajectory': self.trajectories[idx],
            'amplitude': self.amplitudes[idx],
            'frequency': self.frequencies[idx],
            'phase': self.phases[idx]
        }

class ContextEncoder(nn.Module):
    def __init__(self, input_dim, latent_dim, context_dim):
        super().__init__()
        self.input_dim = input_dim
        self.latent_dim = latent_dim
        self.context_dim = context_dim
        
        self.trajectory_encoder = nn.Sequential(
            nn.Linear(input_dim, 256),
            nn.ReLU(),
            nn.Linear(256, 128),
            nn.ReLU(),
        )
        
        self.context_encoder = nn.Sequential(
            nn.Linear(context_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 128),
            nn.ReLU(),
        )
        
        self.fusion = nn.Sequential(
            nn.Linear(128 + 128, 128),
            nn.ReLU(),
        )
        
        self.fc_mu = nn.Linear(128, latent_dim)
        self.fc_logvar = nn.Linear(128, latent_dim)
        
    def forward(self, x, context):
        h_traj = self.trajectory_encoder(x)
        h_context = self.context_encoder(context)
        
        h_fused = self.fusion(torch.cat([h_traj, h_context], dim=1))
        
        mu = self.fc_mu(h_fused)
        logvar = self.fc_logvar(h_fused)
        
        return mu, logvar

class ContextDecoder(nn.Module):
    def __init__(self, latent_dim, output_dim, context_dim):
        super().__init__()
        
        self.context_modulation = nn.Sequential(
            nn.Linear(context_dim, 64),
            nn.ReLU(),
            nn.Linear(64, latent_dim),
        )
        
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 128),
            nn.ReLU(),
            nn.Linear(128, 256),
            nn.ReLU(),
            nn.Linear(256, output_dim)
        )
        
    def forward(self, z, context):
        context_scale = torch.sigmoid(self.context_modulation(context))
        z_modulated = z * context_scale
        
        return self.decoder(z_modulated)

class ContextAwarePINNVAE(nn.Module):
    def __init__(self, input_dim=100, latent_dim=3, context_dim=3):
        super().__init__()
        self.latent_dim = latent_dim
        
        self.encoder = ContextEncoder(input_dim, latent_dim, context_dim)
        self.decoder = ContextDecoder(latent_dim, input_dim, context_dim)
        
    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std
    
    def forward(self, x, context):
        mu, logvar = self.encoder(x, context)
        z = self.reparameterize(mu, logvar)
        x_recon = self.decoder(z, context)
        return x_recon, mu, logvar, z

def compute_physics_loss(x_recon, z, t):
    """
    Physics loss for SHO: x(t) = A * sin(omega * t + phi)
    Compare reconstruction to analytic SHO form
    """
    batch_size = x_recon.shape[0]
    
    # Extract physical parameters from latent space
    # We assume latent dims correspond to amplitude, frequency, phase
    A = torch.abs(z[:, 0:1])      # amplitude (keep positive)
    omega = torch.abs(z[:, 1:2])  # frequency (keep positive)
    phi = z[:, 2:3]               # phase (can be any value)
    
    # Create analytic SHO signal
    t_expanded = t.unsqueeze(0).expand(batch_size, -1)  # [B, T]
    x_analytic = A * torch.sin(omega * t_expanded + phi)
    
    # Loss: reconstruction should match SHO dynamics
    return F.mse_loss(x_recon, x_analytic)

def train_context_aware(model, train_loader, t, epochs=100, lr=1e-3):
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    
    losses = {'total': [], 'recon': [], 'kl': [], 'physics': []}
    
    for epoch in range(epochs):
        model.train()
        epoch_losses = {'total': 0, 'recon': 0, 'kl': 0, 'physics': 0}
        
        for batch in train_loader:
            x = batch['trajectory'].to(dtype=torch.float32, device=device) 
            
            context = torch.stack([
                batch['amplitude'],
                batch['frequency'],
                batch['phase']
            ], dim=1).to(dtype=torch.float32, device=device)
            
            optimizer.zero_grad()
            
            x_recon, mu, logvar, z = model(x, context)
            
            recon_loss = F.mse_loss(x_recon, x)
            
            kl_loss = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp()) / x.shape[0]
            
            physics_loss = compute_physics_loss(x_recon, z, t)
            
            beta_kl = 0.01
            beta_physics = 0.1
            
            total_loss = recon_loss + beta_kl * kl_loss + beta_physics * physics_loss
            
            total_loss.backward()
            optimizer.step()
            
            epoch_losses['total'] += total_loss.item()
            epoch_losses['recon'] += recon_loss.item()
            epoch_losses['kl'] += kl_loss.item()
            epoch_losses['physics'] += physics_loss.item()
        
        for key in epoch_losses:
            epoch_losses[key] /= len(train_loader)
            losses[key].append(epoch_losses[key])
        
        if (epoch + 1) % 20 == 0:
            #plot_model_reconstruction(model, train_loader.dataset, t, epoch + 1)
            print(f"Epoch {epoch+1}/{epochs}")
            print(f"  Total: {epoch_losses['total']:.4f}, Recon: {epoch_losses['recon']:.4f}")
            print(f"  KL: {epoch_losses['kl']:.4f}, Physics: {epoch_losses['physics']:.4f}")
    
    return losses


def analyze_latent_alignment(model, dataset, device='cpu'):
    model.eval()
    
    n_test = 500
    indices = np.random.choice(len(dataset), n_test, replace=False)
    
    latents = []
    true_params = {'amplitude': [], 'frequency': [], 'phase': []}
    
    with torch.no_grad():
        for idx in indices:
            sample = dataset[idx]
            x = torch.FloatTensor(sample['trajectory']).unsqueeze(0).to(device)
            
            context = torch.FloatTensor([
                sample['amplitude'],
                sample['frequency'],
                sample['phase']
            ]).unsqueeze(0).to(device)
            
            mu, _ = model.encoder(x, context)
            latents.append(mu.cpu().numpy()[0])
            
            true_params['amplitude'].append(sample['amplitude'])
            true_params['frequency'].append(sample['frequency'])
            true_params['phase'].append(sample['phase'])
    
    latents = np.array(latents)
    
    fig, axes = plt.subplots(3, 3, figsize=(15, 12))
    param_names = ['amplitude', 'frequency', 'phase']
    
    for i, param_name in enumerate(param_names):
        true_vals = np.array(true_params[param_name])
        
        for j in range(3):
            ax = axes[i, j]
            ax.scatter(true_vals, latents[:, j], alpha=0.5, s=10)
            ax.set_xlabel(f'True {param_name}')
            ax.set_ylabel(f'Latent dim {j}')
            
            corr = np.corrcoef(true_vals, latents[:, j])[0, 1]
            ax.set_title(f'Correlation: {corr:.3f}')
            ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('results/context_aware_latent_alignment.png', dpi=150, bbox_inches='tight')
    plt.close()
    
    print("\nLatent-Physical Parameter Correlations (Context-Aware):")
    for i, param_name in enumerate(param_names):
        true_vals = np.array(true_params[param_name])
        print(f"\n{param_name.capitalize()}:")
        for j in range(3):
            corr = np.corrcoef(true_vals, latents[:, j])[0, 1]
            print(f"  Latent {j}: {corr:.3f}")

if __name__ == "__main__":
    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    print(f"Using device: {device}")
    
    dataset = SHODataset(n_samples=5000)
    train_loader = DataLoader(dataset, batch_size=64, shuffle=True)
    
    t = torch.FloatTensor(dataset.t).to(device)
    
    model = ContextAwarePINNVAE(input_dim=100, latent_dim=3, context_dim=3).to(device)
    
    print("\nTraining Context-Aware PINN-VAE...")
    losses = train_context_aware(model, train_loader, t, epochs=100)
    
    fig, axes = plt.subplots(2, 2, figsize=(12, 8))
    axes = axes.flatten()
    
    for idx, (key, values) in enumerate(losses.items()):
        axes[idx].plot(values)
        axes[idx].set_xlabel('Epoch')
        axes[idx].set_ylabel('Loss')
        axes[idx].set_title(f'{key.capitalize()} Loss')
        axes[idx].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('results/context_aware_training_losses.png', dpi=150, bbox_inches='tight')
    plt.close()
    
    print("\nAnalyzing latent space alignment...")
    analyze_latent_alignment(model, dataset, device)
    
    print("\nContext-Aware PINN-VAE training complete!")
    print("Results saved to results/")
```

##  Intervention-Based PINN-VAE

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
from torch.utils.data import Dataset, DataLoader
import matplotlib.pyplot as plt
from pathlib import Path

import matplotlib.pyplot as plt
import os

results_dir = "results"
os.makedirs(results_dir, exist_ok=True)

#Path("results").mkdir(exist_ok=True)

def get_intervention_loss_schedule(epoch):
    if epoch < 50:
        return {
            'recon': 1.0,
            'kl': 0.001,
            'physics': 0.0,
            'locality': 0.0,
            'sparsity': 0.0,
            'diversity': 0.0
        }
    elif epoch < 150:
        physics_weight = 0.1 + 0.9 * (epoch - 50) / 100
        return {
            'recon': 1.0,
            'kl': 0.001,
            'physics': physics_weight,
            'locality': 0.0,
            'sparsity': 0.0,
            'diversity': 0.0
        }
    else:
        # Gradual warmup + reduced weights
        progress = min((epoch - 150) / 100, 1.0)
        
        return {
            'recon': 1.0,
            'kl': 0.001,
            'physics': 1.0,
            'locality': 0.5 * progress,    # max 0.5 instead of 1.0
            'sparsity': 0.02 * progress,   # max 0.02 instead of 0.05
            'diversity': 0.05 * progress   # max 0.05 instead of 0.1
        }

def plot_model_reconstruction(model, dataset, t, epoch, fixed_idx):
    """
    Plot reconstructions of the same fixed samples across epochs
    """
    model.eval()
    fig, axes = plt.subplots(len(fixed_idx), 1, figsize=(6, 1.5 * len(fixed_idx)), sharex=True)

    with torch.no_grad():
        for i, ax in enumerate(axes):
            sample = dataset[fixed_idx[i]]
            x = torch.FloatTensor(sample['x_original']).unsqueeze(0).to(next(model.parameters()).device)
            x_recon, mu, logvar, z = model(x)
            x = x.squeeze().cpu().numpy()
            x_recon = x_recon.squeeze().cpu().numpy()

            ax.plot(t.cpu().numpy(), x, label="True", linewidth=1)
            ax.plot(t.cpu().numpy(), x_recon, "--", label="Reconstructed", linewidth=1)
            ax.set_ylabel(f"Traj {fixed_idx[i]}")
            ax.legend(fontsize="small", loc="upper right")

    plt.tight_layout()
    fname = os.path.join(results_dir, f"reconstruction_progress_epoch_{epoch:04d}.png")
    plt.savefig(fname, dpi=150, bbox_inches='tight')
    plt.close()
    print(f"Saved {fname}")


class SHOInterventionDataset(Dataset):
    def __init__(self, n_samples=5000, t_points=100, t_max=10.0):
        self.n_samples = n_samples
        self.t = np.linspace(0, t_max, t_points)
        
        self.data = []
        
        for i in range(n_samples):
            A = np.random.uniform(0.5, 3.0)
            w = np.random.uniform(0.5, 3.0)
            phi = np.random.uniform(0, 2*np.pi)
            
            x_original = A * np.sin(w * self.t + phi)
            
            intervention_type = np.random.choice(['amplitude', 'frequency', 'phase'])
            
            if intervention_type == 'amplitude':
                delta_A = np.random.uniform(-0.5, 0.5)
                A_new = np.clip(A + delta_A, 0.5, 3.0)
                x_intervened = A_new * np.sin(w * self.t + phi)
                params_original = [A, w, phi]
                params_intervened = [A_new, w, phi]
                
            elif intervention_type == 'frequency':
                delta_w = np.random.uniform(-0.3, 0.3)
                w_new = np.clip(w + delta_w, 0.5, 3.0)
                x_intervened = A * np.sin(w_new * self.t + phi)
                params_original = [A, w, phi]
                params_intervened = [A, w_new, phi]
                
            else:
                delta_phi = np.random.uniform(-np.pi/2, np.pi/2)
                phi_new = (phi + delta_phi) % (2*np.pi)
                x_intervened = A * np.sin(w * self.t + phi_new)
                params_original = [A, w, phi]
                params_intervened = [A, w, phi_new]
            
            self.data.append({
                'x_original': x_original.astype(np.float32),
                'x_intervened': x_intervened.astype(np.float32),
                'intervention_type': intervention_type,
                'params_original': np.array(params_original, dtype=np.float32),
                'params_intervened': np.array(params_intervened, dtype=np.float32)
            })
    
    def __len__(self):
        return self.n_samples
    
    def __getitem__(self, idx):
        return self.data[idx]

class InterventionEncoder(nn.Module):
    def __init__(self, input_dim, latent_dim):
        super().__init__()
        
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 256),
            nn.ReLU(),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Linear(128, 64),
            nn.ReLU(),
        )
        
        self.fc_mu = nn.Linear(64, latent_dim)
        self.fc_logvar = nn.Linear(64, latent_dim)
        
    def forward(self, x):
        h = self.encoder(x)
        mu = self.fc_mu(h)
        logvar = self.fc_logvar(h)
        return mu, logvar

class InterventionDecoder(nn.Module):
    def __init__(self, latent_dim, output_dim):
        super().__init__()
        
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 128),
            nn.ReLU(),
            nn.Linear(128, 256),
            nn.ReLU(),
            nn.Linear(256, output_dim)
        )
        
    def forward(self, z):
        return self.decoder(z)

class InterventionMaskNetwork(nn.Module):
    def __init__(self, latent_dim, n_intervention_types=3):
        super().__init__()
        self.latent_dim = latent_dim
        self.n_intervention_types = n_intervention_types
        
        self.masks = nn.Parameter(torch.randn(n_intervention_types, latent_dim))
        
    def forward(self, intervention_idx):
        masks = torch.sigmoid(self.masks[intervention_idx])
        return masks

class InterventionPINNVAE(nn.Module):
    def __init__(self, input_dim=100, latent_dim=3):
        super().__init__()
        self.latent_dim = latent_dim
        
        self.encoder = InterventionEncoder(input_dim, latent_dim)
        self.decoder = InterventionDecoder(latent_dim, input_dim)
        self.mask_network = InterventionMaskNetwork(latent_dim, n_intervention_types=3)
        
        self.intervention_to_idx = {
            'amplitude': 0,
            'frequency': 1,
            'phase': 2
        }
        
    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std
    
    def forward(self, x):
        mu, logvar = self.encoder(x)
        z = self.reparameterize(mu, logvar)
        x_recon = self.decoder(z)
        return x_recon, mu, logvar, z

def compute_physics_loss(x, z, t):
    """
    Physics loss for SHO: x(t) = A * sin(omega * t + phi)
    Compare reconstruction to analytic SHO form
    """
    batch_size = x.shape[0]
    
    # Extract physical parameters from latent space
    A = torch.abs(z[:, 0:1])      # amplitude (keep positive)
    omega = torch.abs(z[:, 1:2])  # frequency (keep positive)
    phi = z[:, 2:3]               # phase (can be any value)
    
    # Create analytic SHO signal
    t_expanded = t.unsqueeze(0).expand(batch_size, -1)  # [B, T]
    x_analytic = A * torch.sin(omega * t_expanded + phi)
    
    # Loss: reconstruction should match SHO dynamics
    return F.mse_loss(x, x_analytic)

def compute_intervention_losses(model, x_orig, x_inter, intervention_types):
    x_recon_orig, mu_orig, logvar_orig, z_orig = model(x_orig)
    x_recon_inter, mu_inter, logvar_inter, z_inter = model(x_inter)
    
    intervention_indices = torch.LongTensor([
        model.intervention_to_idx[it] for it in intervention_types
    ]).to(x_orig.device)
    
    masks = model.mask_network(intervention_indices)
    
    delta_z = z_inter - z_orig
    
    locality_loss = torch.mean((delta_z * (1 - masks)) ** 2)
    
    sparsity_loss = torch.mean(torch.sum(masks, dim=1))
    
    consistency_loss = torch.mean((delta_z * masks) ** 2)
    
    all_masks = torch.sigmoid(model.mask_network.masks)
    diversity_matrix = torch.mm(all_masks, all_masks.t())
    mask_norms = torch.norm(all_masks, dim=1, keepdim=True)
    diversity_matrix = diversity_matrix / (mask_norms * mask_norms.t() + 1e-8)
    diversity_matrix = diversity_matrix - torch.eye(3).to(x_orig.device)
    diversity_loss = torch.mean(diversity_matrix ** 2)
    
    return locality_loss, sparsity_loss, consistency_loss, diversity_loss

def train_intervention_based(model, train_loader, t, epochs=100, lr=1e-3):
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    
    losses = {
        'total': [], 'recon': [], 'kl': [], 'physics': [],
        'locality': [], 'sparsity': [], 'diversity': []
    }
    fixed_idx = np.random.choice(len(train_loader.dataset), size=3, replace=False)
    for epoch in range(epochs):
        model.train()
        epoch_losses = {key: 0 for key in losses.keys()}
        
        for batch in train_loader:
            x_orig = batch['x_original']
            x_inter = batch['x_intervened']
            intervention_types = batch['intervention_type']
            
            optimizer.zero_grad()
            
            x_recon_orig, mu_orig, logvar_orig, z_orig = model(x_orig)
            x_recon_inter, mu_inter, logvar_inter, z_inter = model(x_inter)
            
            recon_loss = F.mse_loss(x_recon_orig, x_orig) + F.mse_loss(x_recon_inter, x_inter)
            
            kl_loss = -0.5 * torch.sum(1 + logvar_orig - mu_orig.pow(2) - logvar_orig.exp())
            kl_loss += -0.5 * torch.sum(1 + logvar_inter - mu_inter.pow(2) - logvar_inter.exp())
            kl_loss /= (2 * x_orig.shape[0])
            
            physics_loss = compute_physics_loss(x_recon_orig, z_orig, t)
            physics_loss += compute_physics_loss(x_recon_inter, z_inter, t)
            physics_loss /= 2
            
            locality_loss, sparsity_loss, consistency_loss, diversity_loss = \
                compute_intervention_losses(model, x_orig, x_inter, intervention_types)
            
            weight_dict = get_intervention_loss_schedule(epoch)
            
            beta_recon =  weight_dict['recon']
            beta_kl = weight_dict['kl']          # encourage some structure, but not collapse
            beta_physics = weight_dict['physics']     # make physics ten times louder
            beta_locality =  weight_dict['locality']   # push latent coordinates to move locally under small perturbations
            beta_sparsity = weight_dict['sparsity']    # avoid dense entanglement
            beta_diversity = weight_dict['diversity']   # promote distinct causal directions
            
            total_loss = (beta_recon*recon_loss + 
                         beta_kl * kl_loss + 
                         beta_physics * physics_loss +
                         beta_locality * locality_loss +
                         beta_sparsity * sparsity_loss +
                         beta_diversity * diversity_loss)
            
            total_loss.backward()
            optimizer.step()
            
            epoch_losses['total'] += total_loss.item()
            epoch_losses['recon'] += recon_loss.item()
            epoch_losses['kl'] += kl_loss.item()
            epoch_losses['physics'] += physics_loss.item()
            epoch_losses['locality'] += locality_loss.item()
            epoch_losses['sparsity'] += sparsity_loss.item()
            epoch_losses['diversity'] += diversity_loss.item()
        
        for key in epoch_losses:
            epoch_losses[key] /= len(train_loader)
            losses[key].append(epoch_losses[key])
        
        if (epoch + 1) % 20 == 0:
            plot_model_reconstruction(model, train_loader.dataset, t, epoch + 1, fixed_idx)
            print(f"Epoch {epoch+1}/{epochs}")
            print(f"  Total: {epoch_losses['total']:.4f}, Recon: {epoch_losses['recon']:.4f}")
            print(f"  KL: {epoch_losses['kl']:.4f}, Physics: {epoch_losses['physics']:.4f}")
            print(f"  Locality: {epoch_losses['locality']:.4f}, Sparsity: {epoch_losses['sparsity']:.4f}")
    
    return losses

def analyze_intervention_alignment(model, dataset, device='cpu'):
    model.eval()
    
    n_test = 500
    indices = np.random.choice(len(dataset), n_test, replace=False)
    
    latents = []
    true_params = []
    
    with torch.no_grad():
        for idx in indices:
            sample = dataset[idx]
            x = torch.FloatTensor(sample['x_original']).unsqueeze(0).to(device)
            
            mu, _ = model.encoder(x)
            latents.append(mu.cpu().numpy()[0])
            true_params.append(sample['params_original'])
    
    latents = np.array(latents)
    true_params = np.array(true_params)
    
    fig, axes = plt.subplots(3, 3, figsize=(15, 12))
    param_names = ['Amplitude', 'Frequency', 'Phase']
    
    for i in range(3):
        for j in range(3):
            ax = axes[i, j]
            ax.scatter(true_params[:, i], latents[:, j], alpha=0.5, s=10)
            ax.set_xlabel(f'True {param_names[i]}')
            ax.set_ylabel(f'Latent dim {j}')
            
            corr = np.corrcoef(true_params[:, i], latents[:, j])[0, 1]
            ax.set_title(f'Correlation: {corr:.3f}')
            ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('results/intervention_latent_alignment.png', dpi=150, bbox_inches='tight')
    plt.close()
    
    print("\nLatent-Physical Parameter Correlations (Intervention-Based):")
    for i, param_name in enumerate(param_names):
        print(f"\n{param_name}:")
        for j in range(3):
            corr = np.corrcoef(true_params[:, i], latents[:, j])[0, 1]
            print(f"  Latent {j}: {corr:.3f}")
    
    print("\nLearned Intervention Masks:")
    with torch.no_grad():
        masks = torch.sigmoid(model.mask_network.masks).cpu().numpy()
        intervention_names = ['Amplitude', 'Frequency', 'Phase']
        
        for i, name in enumerate(intervention_names):
            print(f"\n{name} intervention affects:")
            for j in range(3):
                print(f"  Latent {j}: {masks[i, j]:.3f}")




if __name__ == "__main__":
    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    print(f"Using device: {device}")
    
    dataset = SHOInterventionDataset(n_samples=5000)
    train_loader = DataLoader(dataset, batch_size=64, shuffle=True)
    
    t = torch.FloatTensor(dataset.t).to(device)
    
    model = InterventionPINNVAE(input_dim=100, latent_dim=3).to(device)
    
    print("\nTraining Intervention-Based PINN-VAE...")
    losses = train_intervention_based(model, train_loader, t, epochs=300, lr = 1e-3)
    
    fig, axes = plt.subplots(2, 4, figsize=(16, 8))
    axes = axes.flatten()
    
    for idx, (key, values) in enumerate(losses.items()):
        if idx < len(axes):
            axes[idx].plot(values)
            axes[idx].set_xlabel('Epoch')
            axes[idx].set_ylabel('Loss')
            axes[idx].set_title(f'{key.capitalize()} Loss')
            axes[idx].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('results/intervention_training_losses.png', dpi=150, bbox_inches='tight')
    plt.close()
    
    
    print("\nAnalyzing latent space alignment...")
    analyze_intervention_alignment(model, dataset, device)
    
    print("\nIntervention-Based PINN-VAE training complete!")
    print("Results saved to results/")
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

Path("results").mkdir(exist_ok=True)

def compute_disentanglement_score(latents, true_params):
    """
    Compute disentanglement metric: how well each latent dimension 
    aligns with a single physical parameter.
    
    Higher score = better disentanglement
    """
    n_latent = latents.shape[1]
    n_params = true_params.shape[1]
    
    corr_matrix = np.zeros((n_params, n_latent))
    
    for i in range(n_params):
        for j in range(n_latent):
            corr_matrix[i, j] = np.abs(np.corrcoef(true_params[:, i], latents[:, j])[0, 1])
    
    max_corr_per_latent = np.max(corr_matrix, axis=0)
    disentanglement_score = np.mean(max_corr_per_latent)
    
    assignment = np.argmax(corr_matrix, axis=0)
    
    return disentanglement_score, corr_matrix, assignment

def analyze_model(model, dataset, device, model_type):
    """Analyze trained model and extract latent representations"""
    model.eval()
    
    n_test = 500
    indices = np.random.choice(len(dataset), n_test, replace=False)
    
    latents = []
    true_params = []
    
    with torch.no_grad():
        for idx in indices:
            sample = dataset[idx]
            
            if model_type == "context_aware":
                x = torch.FloatTensor(sample['trajectory']).unsqueeze(0).to(device)
                context = torch.FloatTensor([
                    sample['amplitude'],
                    sample['frequency'],
                    sample['phase']
                ]).unsqueeze(0).to(device)
                
                mu, _ = model.encoder(x, context)
                latents.append(mu.cpu().numpy()[0])
                true_params.append([
                    sample['amplitude'],
                    sample['frequency'],
                    sample['phase']
                ])
            else:
                x = torch.FloatTensor(sample['x_original']).unsqueeze(0).to(device)
                mu, _ = model.encoder(x)
                latents.append(mu.cpu().numpy()[0])
                true_params.append(sample['params_original'])
    
    latents = np.array(latents)
    true_params = np.array(true_params)
    
    return latents, true_params

def compare_approaches(context_model, intervention_model, 
                      context_dataset, intervention_dataset, device):
    """Compare both approaches side by side"""
    
    print("\n" + "="*70)
    print("COMPARISON: Context-Aware vs Intervention-Based PINN-VAE")
    print("="*70)
    
    latents_context, params_context = analyze_model(
        context_model, context_dataset, device, "context_aware"
    )
    
    latents_intervention, params_intervention = analyze_model(
        intervention_model, intervention_dataset, device, "intervention_based"
    )
    
    score_context, corr_context, assign_context = compute_disentanglement_score(
        latents_context, params_context
    )
    
    score_intervention, corr_intervention, assign_intervention = compute_disentanglement_score(
        latents_intervention, params_intervention
    )
    
    print("\n1. DISENTANGLEMENT SCORES (higher is better)")
    print("-" * 70)
    print(f"Context-Aware:      {score_context:.4f}")
    print(f"Intervention-Based: {score_intervention:.4f}")
    
    param_names = ['Amplitude', 'Frequency', 'Phase']
    
    print("\n2. CORRELATION MATRICES")
    print("-" * 70)
    
    print("\nContext-Aware (Physical Parameter -> Latent Dimension):")
    print("           Latent 0  Latent 1  Latent 2")
    for i, name in enumerate(param_names):
        print(f"{name:10s} {corr_context[i, 0]:8.3f}  {corr_context[i, 1]:8.3f}  {corr_context[i, 2]:8.3f}")
    
    print("\nIntervention-Based (Physical Parameter -> Latent Dimension):")
    print("           Latent 0  Latent 1  Latent 2")
    for i, name in enumerate(param_names):
        print(f"{name:10s} {corr_intervention[i, 0]:8.3f}  {corr_intervention[i, 1]:8.3f}  {corr_intervention[i, 2]:8.3f}")
    
    print("\n3. LATENT DIMENSION ASSIGNMENTS")
    print("-" * 70)
    
    print("\nContext-Aware:")
    for j in range(3):
        assigned_param = param_names[assign_context[j]]
        max_corr = corr_context[assign_context[j], j]
        print(f"  Latent {j} -> {assigned_param} (corr: {max_corr:.3f})")
    
    print("\nIntervention-Based:")
    for j in range(3):
        assigned_param = param_names[assign_intervention[j]]
        max_corr = corr_intervention[assign_intervention[j], j]
        print(f"  Latent {j} -> {assigned_param} (corr: {max_corr:.3f})")
    
    if hasattr(intervention_model, 'mask_network'):
        print("\n4. LEARNED INTERVENTION MASKS (Intervention-Based only)")
        print("-" * 70)
        with torch.no_grad():
            masks = torch.sigmoid(intervention_model.mask_network.masks).cpu().numpy()
            
            print("\nMask values (closer to 1.0 = dimension affected by intervention):")
            print("              Latent 0  Latent 1  Latent 2")
            for i, name in enumerate(param_names):
                print(f"{name:10s}  {masks[i, 0]:8.3f}  {masks[i, 1]:8.3f}  {masks[i, 2]:8.3f}")
            
            print("\nInterpretation:")
            for i, name in enumerate(param_names):
                affected_dims = np.where(masks[i] > 0.5)[0]
                if len(affected_dims) > 0:
                    dims_str = ", ".join([f"Latent {d}" for d in affected_dims])
                    print(f"  {name} intervention primarily affects: {dims_str}")
                else:
                    print(f"  {name} intervention: no strong effects learned")
    
    fig, axes = plt.subplots(2, 3, figsize=(15, 10))
    
    for i, param_name in enumerate(param_names):
        ax_context = axes[0, i]
        ax_intervention = axes[1, i]
        
        best_latent_context = np.argmax(corr_context[i])
        best_latent_intervention = np.argmax(corr_intervention[i])
        
        ax_context.scatter(params_context[:, i], latents_context[:, best_latent_context], 
                          alpha=0.5, s=10)
        ax_context.set_xlabel(f'True {param_name}')
        ax_context.set_ylabel(f'Latent {best_latent_context}')
        corr = corr_context[i, best_latent_context]
        ax_context.set_title(f'Context-Aware (r={corr:.3f})')
        ax_context.grid(True, alpha=0.3)
        
        ax_intervention.scatter(params_intervention[:, i], latents_intervention[:, best_latent_intervention],
                               alpha=0.5, s=10)
        ax_intervention.set_xlabel(f'True {param_name}')
        ax_intervention.set_ylabel(f'Latent {best_latent_intervention}')
        corr = corr_intervention[i, best_latent_intervention]
        ax_intervention.set_title(f'Intervention-Based (r={corr:.3f})')
        ax_intervention.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('results/approach_comparison.png', dpi=150, bbox_inches='tight')
    plt.close()
    
    print("\n5. KEY DIFFERENCES")
    print("-" * 70)
    print("""
Context-Aware Approach:
  - Uses physical parameters as additional input (context)
  - Context modulates the encoding process
  - May learn task-specific representations
  - Relies on having parameter labels during training
  
Intervention-Based Approach:
  - Learns from pairs of trajectories (before/after intervention)
  - Discovers which dimensions change with each intervention
  - Enforces sparsity: interventions should affect few dimensions
  - Does NOT need exact parameter values, only intervention types
  - More closely mimics experimental science workflow
    """)
    
    print("\n" + "="*70)
    print("Analysis complete! Check results/ for visualizations.")
    print("="*70)

if __name__ == "__main__":
    print("""
This comparison script should be run AFTER training both models.
    
Usage:
    1. First run: context_aware_pinn_vae.py
    2. Then run: intervention_pinn_vae.py
    3. Finally run this script with the trained models
    
For demonstration purposes, we'll show the structure of the comparison.
Load your trained models and datasets, then call compare_approaches().
    """)
    
    print("\nExample usage:")
    print("""
    from context_aware_pinn_vae import ContextAwarePINNVAE, SHODataset
    from intervention_pinn_vae import InterventionPINNVAE, SHOInterventionDataset
    
    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    
    context_model = ContextAwarePINNVAE().to(device)
    context_model.load_state_dict(torch.load('context_model.pth'))
    
    intervention_model = InterventionPINNVAE().to(device)
    intervention_model.load_state_dict(torch.load('intervention_model.pth'))
    
    context_dataset = SHODataset(n_samples=1000)
    intervention_dataset = SHOInterventionDataset(n_samples=1000)
    
    compare_approaches(context_model, intervention_model, 
                      context_dataset, intervention_dataset, device)
    """)
```
